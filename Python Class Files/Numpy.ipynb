# Numpy
NumPy (Numerical Python) is the fundamental open-source library for
scientific computing in Python, providing support for large, multi-dimensional
arrays and matrices.

## Key Features of NumPy

Ndarray Object: The core is the ndarray, a fast, flexible, homogeneous ,dimensional array object that enables efficient storage and manipulation of numerical data.

Performance: Written partially in C or C++, NumPy operations are significantly faster than standard Python lists, making it ideal for heavy computation.

Mathematical Functions: Includes tools for linear algebra, Fourier transforms, random number generation, and  ophisticated sorting/selection.

Fixed-Size Arrays: Unlike Python lists, NumPy arrays have a fixed size at creation, optimizing memory usage and processing speed.

## Change data type of an array

```python
arr = np.array([1,2,3,4])
print(arr)

arr = arr.astype(float)
print(arr)

arr = arr.astype(str)
print(arr)
```

```python
arr = np.array([1.453, 2.345, 3.567,8.7665,12.765])
print(arr)

arr = arr.astype(int)
print(arr)
```
