# np.where

```python
import numpy as np

print(sales)

np.where(sales>=500 , "good", "bad")
```

# apply method

```python
print(sales)

def find_category(amount):
    if amount<400:
        return "Low"
    elif amount < 700:
        return "Mid"
    else:
        return "High"

sales.apply(find_category)
```

```python
sales.apply(lambda amount : amount*0.05 if amount<=500 else amount*0.10)
```
