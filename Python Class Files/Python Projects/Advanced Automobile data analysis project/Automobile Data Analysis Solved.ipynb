### 2. Find the Most Powerful Car per Company
For each car company, find the car(s) with the maximum horsepower and display their full details.

```python
final_df = pd.DataFrame()
for i in df['company'].unique():
    data = df[df['company'] == i]
    max_horsepower = data['horsepower'].max()
    final_df =  pd.concat([final_df,data[data['horsepower'] == max_horsepower]])

final_df.sort_values(by='company')
```
