# Nested If

```python
age = int(input("Whate is your age"))

if age >= 18:
    print("Adult")
    budget = int(input("What is you budget ?"))
    if budget >= 500000:
        print("you can get a car")
    else:
        print("you can get a Bike")

else:
    print("Child")
    marks = int(input("what are you school marks"))
    if marks>= 70:
        print("you get a Cycle")
    else:
        print("You get a Board game Ludo")
```

```python
# Write a program to check whether a year is a leap year or not.

year = int(input('year -'))

if year % 100 == 0:
    if year % 400==0:
        print("leap year")
    else:
        print("normal year")

else:
    if year % 4 == 0:
        print("leap year")
    else:
        print("Normal year")
```

```python
year = 2013
"leap year" if year%400==0 or (year%100 != 0 and year%4 ==0) else "Normal year"
```

# ternary operator (Shortcut of if function)

```python
age = 13

"can vote" if age >= 18 else "can not vote"
```

```python
# Give discount 12% only if the sales is above 5000 else 0

sales = 6700
discount = sales*12/100 if sales >= 5000 else 0

print(discount)
```

```python
# Find profit and loss
amount = 600
print("profit") if amount>0 else print("Loss")
```

```python
# Find the discount rate on the basis of sales
# below 1200 - 5%
# 1200 - 4000 - 8%
# above 4000 - 12%

sales = 3210

print('5%') if sales < 1200 else print("8%") if sales < 4000 else print("12%")
```
