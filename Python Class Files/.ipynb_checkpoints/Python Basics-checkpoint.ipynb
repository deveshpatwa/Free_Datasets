# Variables
- a variable is a name that you use to store data or a value in memory so that you can use
it later in your program.
- It acts like a container that holds some information
- for example, a number, text, or list —
- allows you to refer to that data by name instead of writing it directly every
time.

```python
45000 * 12 /100
```

## Rules for naming variables
- Variable names can include letters(a-z,A-Z), numbers(0-9), and
underscores (_).
- They must start with a letter or underscore, not a number.
- Variable names are case-sensitive (Name and name are different).
- Don’t use Python keywords (like if, for, class, etc.) as variable names.

```python
# print simple interest using variables -  amount , year and interest rate?
# And then show compound interest of same

amount = 12000
interest_rate = 7/100
year = 4

print(amount * interest_rate * year)
```
