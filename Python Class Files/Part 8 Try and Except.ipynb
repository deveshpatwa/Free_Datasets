# Try and Except

Try to run this code. If an error happens, don't crash the program—instead, do something else.

```python
# try:
#     Code that might cause an error
# except:
#     Runs only if an error occurs
```

```python
try:
    print(10 / 0)
except:
    print("can not divide by 0")
print("imp code")
```

```python
try:
    print("Line 1")
    print(10 / 0)
    print("Line 3")
except:
    print("Error!")
```

```python
try:
    number = int(input("enter a number only"))
    print(number)
except ValueError:
    print("Please enter a valid number.")
```

```python
try:
    num = int(input("Enter number: "))
    print(100 / num)

except ValueError:
    print("Please enter e number only.")

except ZeroDivisionError:
    print("Cannot divide by zero.")
```

```python
try:
    num = int(input("Enter a number: "))
except :
    print("please enter a number")
else:
    print("The number was -", num)
```

```python
try:
    num = int(input("Enter a number: "))
    print(num)
except:
    print("please enter a number")
finally:
    print("Always runs")
```
