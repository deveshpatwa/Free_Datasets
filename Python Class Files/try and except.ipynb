# Try and Except

Try to run this code. If an error happens, don't crash the program—instead, do something else.

```python
# try:
#     Code that might cause an error
# except:
#     Runs only if an error occurs
```

```python
10/0
```

```python
try:
    print(10 / 2)
except:
    print("An error occurred.")

print("Program End")
```

```python
try:
    print("Line 1")
    print(10 / 0)
    print("Line 3")
except:
    print("Error!")
```

```python
try:
    number = int(input("enter a number only"))
except ValueError:
    print("Please enter a valid number.")
```

```python

try:
    num = int(input("Enter number: "))
    print(100 / num)

except ValueError:
    print("You entered text.")

except ZeroDivisionError:
    print("Cannot divide by zero.")
```

```python
try:
    num = int(input("Enter a number: "))
except ValueError:
    print("Invalid input.")
else:
    print("You entered:", num)
```

```python
try:
    print(10 / 0)
except:
    print("Error")
finally:
    print("Always runs")
```
