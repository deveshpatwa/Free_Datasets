A **mutable object** is one whose data can be modiﬁed in place without creating a new object.
This means the object’s identity (memory address) stays the same, but its contents can
change. Common mutable objects in Python include lists, dictionaries, sets, and bytearrays.
For example, if you have a list and you change one of its elements, the same list object is
updated rather than creating a new list.

An **immutable object** is one whose value cannot be changed after it is created. If you try to
modify an immutable object, Python creates a new object instead of changing the existing
one. Common immutable objects include integers, ﬂoats, strings, tuples, and frozensets. For
example, when you change a string, Python actually creates a new string rather than
modifying the original one.

# Set

- A set is a collection data type used to store multiple unique items in a single variable.
- Sets are unordered, unindexed, and do not allow duplicate values.
- You create a set using curly braces { } or the set() function.
- Unordered – The items don’t have a ﬁxed position or index.
- Unindexed – You can’t access elements using indexes like lists.
- No duplicates – Duplicate items are automatically removed.
- Mutable – You can add or remove elements, but cannot change existing ones.
- Heterogeneous – Can store different data types (numbers, strings, etc.).
- A set will store only immutable objects

```python
s = {4,6,5,3}
print(s)
```

```python
s = {4,6,5,3,4,4,3}
print(s)
```

```python
s = {23,8.65, "hello", (5,4,3)}
print(s)
```

```python
s = {23,8.65, "hello", (5,4,3),[54,76,89]}
print(s)
```

```python
s = {3,4,5,6}
s.add(7)
print(s)
```

```python
s = {3,4,5,6}
s.update([8,0,4,3])
print(s)
```

```python
s = {3,4,5,6}
s.remove(9)
print(s)
```
