## Import Libraries

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
# seaborn themes
sns.set_theme(style="darkgrid")
```

## Load Sample Dataset

```python
car = pd.read_csv("car_average.csv")
car.head()
```

```python
# get free dataset from this function
sns.get_dataset_names()
```

```python
tips = sns.load_dataset("tips")
tips.head()
```

```python
store = pd.read_csv("store.csv")
store.head()
```

## Histogram

```python
sns.histplot(data=tips,x='tip')
plt.show()
```

```python
sns.histplot(data=tips,x='tip',hue="sex")
plt.show()
```

```python
sns.histplot(data=car,x="horsepower",hue="origin")
plt.show()
```

## Scatter Plot

```python
sns.scatterplot(data=car,x="horsepower", y="mpg",color='gray')
plt.show()
```

## Line Plot

```python
sns.lineplot(data=tips,x="size", y="tip" )
plt.show()
```

```python
# store.info()
store['order_date'] = pd.to_datetime(store['order_date'])
store['month'] = store['order_date'].dt.month
```

```python
# estimator = sum , min, max, mean etc
sns.lineplot(data=store,x='month',y="sales",estimator="sum")
plt.ylabel("Sales in $")
plt.show()
```

## Bar Plot

```python
sns.barplot(data=tips,x="size", y="tip" )
plt.show()
```

```python
sns.barplot(data=tips,x="size", y="tip",estimator="sum",hue="sex" )
plt.show()
```

```python
sns.barplot(data=tips,x="day", y="total_bill",estimator="sum")
plt.show()
```

```python
# Show sub_category wise total sales from store
data=  store.groupby(['category','sub_category'])['sales'].sum().reset_index().sort_values(by=['category','sales'],ascending=[True,False])
```

```python
data
```

```python
plt.figure(figsize=(8,6))
sns.barplot(data=data,y='sub_category',x='sales',estimator="sum",hue='category')
plt.show()
```

## Heatmap

```python
corr = car.corr(numeric_only=True)
print(corr)
```
