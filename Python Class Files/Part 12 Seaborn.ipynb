## Import Libraries

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
# seaborn themes
sns.set_theme(style="darkgrid")
```

## Load Sample Dataset

```python
car = pd.read_csv("car_average.csv")
car.head()
```

```python
# get free dataset from this function
sns.get_dataset_names()
```

```python
tips = sns.load_dataset("tips")
tips.head()
```

```python
store = pd.read_csv("store.csv")
store.head()
```

## Histogram

```python
sns.histplot(data=tips,x='tip')
plt.show()
```

```python
sns.histplot(data=tips,x='tip',hue="sex")
plt.show()
```

```python
sns.histplot(data=tips,x='tip',kde=True,hue="sex")
plt.show()
```

## Scatter Plot

```python
sns.scatterplot(data=car,x="horsepower", y="mpg",color='green')
plt.show()
```

```python
sns.scatterplot(data=car,x="horsepower", y="mpg",hue="cylinders")
plt.show()
```

```python
# palette option - viridis , plasma, inferno,cividis, rocket
sns.scatterplot(data=car,x="horsepower", y="mpg",hue="origin",palette='viridis')
plt.show()
```

## Line Plot

```python
sns.lineplot(data=tips,x="size", y="tip" )
plt.show()
```

```python
# store.info()
store['order_date'] = pd.to_datetime(store['order_date'])
store['month'] = store['order_date'].dt.month
```

```python

sns.lineplot(data=store,x='month',y="sales",estimator="sum")
plt.show()
```

## Bar Plot

```python
sns.barplot(data=tips,x="day", y="total_bill",estimator="sum")
plt.show()
```

```python
# Show category wise total sales from store
```

## KDE Plot

```python
sns.kdeplot(x=tips["total_bill"])
plt.show()
```

```python
sns.kdeplot(data=store,x='qty')
plt.show()
```

## Box Plot

```python
sns.boxplot(data=tips,x='total_bill')
plt.show()
```

```python
sns.boxplot(data=tips,x='total_bill',y='sex')
plt.show()
```

```python
car['cylinders'] = car['cylinders'].astype(str)
sns.boxplot(data=car,x='mpg',y='cylinders')
plt.show()
```

```python
sns.boxplot(x="day", y="total_bill", data=tips)
plt.show()
```

## Violin Plot

```python
sns.violinplot(data=tips, y="total_bill")
plt.show()
```

```python
sns.violinplot(data=car,x='mpg',y='cylinders')
plt.show()
```

## Count Plot

```python
sns.countplot(x="day", data=tips)
plt.show()
```

## Heatmap

```python
corr = tips.corr(numeric_only=True)
print(corr)
```

```python
sns.heatmap(corr)
plt.show()
```

```python
sns.heatmap(corr, annot=True)
plt.show()
```

```python
corr = car.corr(numeric_only=True)
print(corr)
```

```python
sns.heatmap(corr,annot=True)
plt.show()
```

## Pair Plot

```python
sns.pairplot(tips)
plt.show()
```

```python
sns.pairplot(car)
plt.show()
```

```python
sns.pairplot(car,hue="origin")
plt.show()
```
