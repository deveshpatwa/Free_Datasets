## Import Libraries

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

## Load Sample Dataset

```python
car = pd.read_csv("car_average.csv")
car.head()
```

```python
tips = sns.load_dataset("tips")
tips.head()
```

```python
store = pd.read_csv("store.csv")
store.head()
```

```python
sns.histplot(data=tips,x='tip')
```

```python
# seaborn themes
sns.set_theme(style="darkgrid")
```

```python
sns.histplot(data=tips,x='tip',hue="sex")
```

```python
sns.histplot(data=tips,x='tip',kde=True,hue="sex")
```

## Scatter Plot

```python
sns.scatterplot(data=car,x="horsepower", y="mpg")
plt.show()
```

```python
sns.scatterplot(data=car,x="horsepower", y="mpg",hue="cylinders")
plt.show()
```

```python
sns.scatterplot(data=car,x="horsepower", y="mpg",hue="origin")
plt.show()
```

## Line Plot

```python
sns.lineplot(x="size", y="tip", data=tips)
plt.show()
```

```python
# store.info()
store['order_date'] = pd.to_datetime(store['order_date'])
store['month'] = store['order_date'].dt.month
```

```python

sns.lineplot(data=store,x='month',y="sales",estimator="sum")
plt.show()
```

## Bar Plot

```python
sns.barplot(x="day", y="total_bill", data=tips,estimator="mean")
plt.show()
```

## Histogram

```python
sns.histplot(tips["total_bill"],kde=True)

plt.show()
```

## KDE Plot

```python
sns.kdeplot(tips["total_bill"])

plt.show()
```

## Box Plot

```python
sns.boxplot(data=tips,x='total_bill',y='sex')
```

```python
sns.boxplot(x="day", y="total_bill", data=tips)

plt.show()
```

## Violin Plot

```python
sns.violinplot(x="day", y="total_bill", data=tips)

plt.show()
```

## Count Plot

```python
sns.countplot(x="day", data=tips)

plt.show()
```

## Heatmap

```python
corr = tips.corr(numeric_only=True)

sns.heatmap(corr, annot=True)

plt.show()
```

## Pair Plot

```python
sns.pairplot(tips)

plt.show()
```

## Change Style

```python
sns.set_style("darkgrid")

sns.barplot(x="day", y="total_bill", data=tips)

plt.show()
```

## Change Figure Size

```python
plt.figure(figsize=(10,5))

sns.barplot(x="day", y="total_bill", data=tips)

plt.show()
```

## Add Title and Labels

```python
sns.barplot(x="day", y="total_bill", data=tips)

plt.title("Average Bill by Day")
plt.xlabel("Day")
plt.ylabel("Bill Amount")

plt.show()
```
