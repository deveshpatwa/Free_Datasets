```python
import random as r
import time as t

count = 0
start = t.time()
for i in range(10):
    num1 = r.randint(1,20)
    num2 = r.randint(1,20)
    print(f"what is {num1} + {num2}")
    user = int(input("Ans -"))
    if user == num1+num2:
        count += 1

end = t.time()
print(f"correct ans = {count}/10 and you took {round(end-start,2)} sec.")
```

# Welcome to the "Guess My Number" Game!
In this Python program, you’ll write a simple and fun number guessing game against the
computer.
How the Game Works:
- The computer will randomly select a number between 1 and 100.
- Your task is to guess the number chosen by the computer.
- After each guess:
- If your guess is correct, the computer will congratulate you.
- If your guess is incorrect, the computer will tell you whether to guess a higher or
lower number.
- The computer will keep track of the number of attempts you make.
- The game will continue until you guess the correct number.
- Once you find the correct number, the computer will display a congratulatory message
along with the
- total number of attempts you took
