```python
import numpy  as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# data file

df = pd.read_csv(r"C:\Users\deves\Documents\GitHub\Free_Datasets\Data Sets\store.csv")
```

```python
df.head()
```

```python
# bar chart

df.groupby("region")['sales'].sum().plot(kind="bar")
```

```python
# barh chart

df.groupby("region")['sales'].sum().plot(kind="barh")
```

```python
# barh chart formating

plt.figure(figsize=(9,6))

df.groupby("sub_category")['sales'].sum().sort_values().plot(kind="barh",color="gray")

plt.title("Sub category wise total sales",size=15,color="blue",fontweight="bold")
plt.xlabel("Sales")
plt.ylabel("Sub category")
plt.show()
```

```python
# Pie chart
df.category.value_counts().plot(kind="pie")

plt.show()
```

```python
# Line chart
df.info()
```

```python
df['order_date'] = pd.to_datetime(df['order_date'])
df['year'] = df['order_date'].dt.year
df['month'] = df['order_date'].dt.month
```

```python
df.head()
```

```python
plt.figure(figsize=(12,4))

df.groupby(['year','month'])['sales'].sum().plot(kind='line')
plt.show()
```

```python
plt.figure(figsize=(12,4))

df.groupby(['year','month'])['sales'].sum().plot(kind='area')
plt.show()
```

```python
profit =  df[['order_date','profit']].sort_values(by='order_date').reset_index()['profit']
```

```python
profit.cumsum().plot()
plt.show()
```

# Seaborn

```python
import seaborn as sns
```

```python
# set theme
sns.set_theme(style='darkgrid')
```

## Hist plot

```python
sns.histplot(data=car,x='horsepower',color="black",kde=True)

plt.title("Histogram",size=15,fontweight="bold")
plt.show()
```

# scatter plot

```python
sns.scatterplot(data=car,x='horsepower',y='mpg',hue="cylinders")
plt.show()
```

# Regplot

```python
sns.regplot(data=car,x="horsepower",y="mpg")
plt.show()
```

# Box Plot

```python
sns.boxplot(data=car,x="horsepower",y='origin')
plt.show()
```
