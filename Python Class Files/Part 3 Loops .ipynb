# Nested for loop

```python
for name in ['rohan','kartik','kunal','mohit']:
    for product in ['pen','car','bike','shoes']:
        print(f'{name} - {product}')
```

# While Loop

```python
num = 5

while num<=100:
    print(num)
    num = int(input())

print("End of loop")
```

```python
# Write a function to take password from the user if the user gives a wrong password retake it 

password = input("enter your password")

while password != "abcd":
    print("wrong password")
    password = input("enter correct password")

print("welcome")
```
