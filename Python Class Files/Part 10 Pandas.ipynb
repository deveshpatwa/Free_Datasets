```python
# Create a new column for sales category low Medium and High?
def sales_cat(amount):
    if amount <= 100:
        return "Low"
    elif amount <= 1000:
        return "Mid"
    else:
        return "High"

df['sales_category'] = df['sales'].apply(sales_cat)
```
