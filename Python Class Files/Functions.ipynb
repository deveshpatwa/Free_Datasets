# User Defined Functions

- User-defined functions in Python are the functions that you create yourself to perform specific tasks.
- They help make your code more organized, reusable, and easier to understand.
- You define a function using the def keyword, give it a name, optionally add parameters, and then write the code that runs when the function is called.

```python
# discount(amount,rate)
def discount(amount,rate):
    disc = amount * rate / 100
    return disc

discount(5000,4.8)
```

```python
# even_odd(num)
def even_odd(num):
    if num%2==0:
        return "Even"
    else:
        return "Odd"

even_odd(6)
```

# Types of variables in a function
**global and a local variable**
- a local variable is a variable that is created inside a function and can be used only within that function. It exists only while the function is running, and once the function finishes execution, the local variable is destroyed. This means you cannot access a local variable outside the function in which it is defined.
- A global variable, on the other hand, is a variable that is defined outside of all functions, usually at the top level of a program. A global variable can be accessed from anywhere in the code, including inside functions.

```python
a = 45

def test():
    b = 20
    print(a)
    print(b)

test()
```

```python
print(a)
print(b)
```

```python
a = 111

def test():
    global b
    b = 445
    print(a)
    print(b)

test()
```

```python
print(a)
print(b)
```

# Doc String
A docstring (documentation string) in Python is a string literal used to explain the purpose
and behavior of modules, classes, functions, or methods. Defined using triple quotes (""")
immediately following the object's definition, it becomes part of the code's metadata,
making it accessible at runtime via __doc__ or help().

```python
# discount(amount,rate)
def discount(amount,rate):
    """ This function will calculate the discount by using amount and rate given by the user"""
    disc = amount * rate / 100
    return disc

discount(5000,4.8)
```

```python
help(discount)
```

```python
discount.__doc__
```

```python
help(print)
```

# Types of parameters

```python
# Positional Parameters 

def greet(name,age):
    print(f"your name is {name} and your age is {age}")

greet("rohan",34)
greet(34,"rohan")
```

```python
# Keyword Parameters

def greet(name,age):
    print(f"your name is {name} and your age is {age}")

greet(age=34,name="rohan")
```

```python
# Default Parameters

def discount(amount,rate=12):
    disc = amount * rate / 100
    return disc

discount(5000)
```

```python
# Arbitrary parameters: Use an asterisk * before a parameter name to accept any number of positional arguments. The function receives them as a tuple.

a = 4,5,2,8,5
print(a)
```

```python
def total(*a):
    total_of_numbers = 0
    for i in a:
        total_of_numbers += i
    return total_of_numbers

total(4,5,6,8,12,45,76,23)
```

```python
# Arbitrary Keyword Parameters (**kwargs)
def people(**info):
    print(info)
    print(type(info))

people(name='rohan',age=45)
```

```python
def people(**info):
    for i in info.keys():
        print(f"My {i} is {info[i]}")

people(name='rohan',age=45)
```

```python
people(name='Kunal',height=5.9,weight=78)
```

# Recursion Function

```python
def age(num):
    if num>=100:
        new_age = int(input("Enter a correct age"))
        age(new_age)
    else:
        print(f"Your age is {num}")

age(450)
```
