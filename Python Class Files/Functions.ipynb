# Function

a function is a block of code that performs a specific task and can be reused whenever
needed. It helps make programs shorter, cleaner, and easier to manage. A function is defined
using the def keyword followed by the function name, parentheses (which may include
parameters), and a colon. The code inside the function is indented.
Types of functions
- Built-in Functions
- User-defined Functions

# Built-in Functions

```python
# sorted
a = [5,9,2,6,3,8]
sorted(a)
```

```python
a = (5,9,2,6,3,8)
sorted(a)
```

```python
a = {5,9,2,6,3,8}
sorted(a)
```

```python
a = [5,9,2,6,3,8]
sorted(a,reverse=True)
```

```python
# Zip
sub = ['maths','english','hindi','science']
marks = [78,99,56,67]
zip(sub,marks)
```
